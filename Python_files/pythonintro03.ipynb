Topic 53 | NumPy Random Numbers, Reproducibility & Performance Tips
This topic highlights how to generate random numbers using NumPy’s Generator and apply seeding for reproducibility. It also compares vectorized operations with loops and includes a coin flip simulation.

```python
import numpy as np
rng = np.random.default_rng() # Create a random number generator
flips = rng.integers(0, 2, size=100000) # Generate 100,000 random integers (0 or 1)  0 = tails, 1 = heads
prob_heads = np.mean(flips) # Calculate the proportion of heads (1s)
print("Estimated Probability of Heads:", prob_heads) # Output: Approximately 0.5 (varies with each run)
```

Topic 56 | Pandas Indexing & Selection - loc & iloc
This topic explains how to select data using .loc and .iloc. It covers boolean filtering and best practices for safe and clear data selection.

```python
data = {'Name':['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
        'Age':[25, 30, 22, 35, 28],
        'City':['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix'],
        'Salary':[70000, 80000, 60000, 90000, 75000]}
df = pd.DataFrame(data)
print("Original DataFrame:\n", df)
```

```python
# label-based indexing with loc
subset_loc = df.loc[1:3, ['Name', 'City']]  # Select rows 1 to 2 and columns 'Name' and 'City'
print("Subset using loc:\n", subset_loc)
```

```python
# integer position-based indexing with iloc
subset_iloc = df.iloc[1:3, [0, 2]]  # Select rows 1 to 2 and columns at index 0 and 2
print("Subset using iloc:\n", subset_iloc)
```

```python
# Boolean filtering
high_salary = df[df['Salary'] > 75000]  # Filter rows where Salary is greater than 75000
print("Employees with high salary:\n", high_salary)
```

```python
# Safe assignment using iloc
df.loc[df['Age'] < 30, 'Salary'] = 65000  # Set Salary to 65000 for employees younger than 30
print("Low age employees with updated salary:\n", df.loc[df['Age'] < 30, 'Salary'])
```

Topic 62 | Merging & Joining DataFrames
This topic explains how to merge and join DataFrames using different join types. It also covers concatenation and handling column conflicts with suffixes.

```python
import pandas as pd
```

```python
customers = pd.DataFrame({
    'CustomerID': [1, 2, 3, 4, 5],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva']
})

orders = pd.DataFrame({
    'OrderID': [101, 102, 103, 104, 105],
    'CustomerID': [1, 2, 1, 3, 2],
    'Amount': [250, 150, 300, 200, 100]
})
```

```python
df_inner = pd.merge(customers, orders, on='CustomerID', how='inner')  # Inner join on 'CustomerID'
print("Inner Join:\n", df_inner)
```

```python
df_left = pd.merge(customers, orders, on='CustomerID', how='left')  # Left join on 'CustomerID'
print("Left Join:\n", df_left)
```

```python
df_right = pd.merge(customers, orders, on='CustomerID', how='right')  # Right join on 'CustomerID'
print("Right Join:\n", df_right)
```

```python
df_outer = pd.merge(customers, orders, on='CustomerID', how='outer')  # Outer join on 'CustomerID'
print("Outer Join:\n", df_outer)
```

```python
df_concat = pd.concat([customers, customers])  # Concatenate along rows (axis=0)
print("\nConcatenated Rows:\n", df_concat)
```

Topic 63 | Applying Functions - apply, map & pipe
This topic covers applying custom and built in functions using apply, map, and pipe. It also explains performance considerations and method chaining.

```python
import pandas as pd
df = pd.read_csv("ecommerce_sales_analytics_5000.csv", parse_dates=["order_date"])
df.info()  # Print information about the DataFrame
```

```python
df['Payment_Type'] = df['payment_method'].map({
    'Card': 'Digital',
    'Wallet': 'Digital',
    'COD': 'Cash',
})

df['Payment_Type'].unique()  # Display unique values in the 'Payment_Type' column
```

```python
df[['Payment_Type', 'payment_method']].head()  # Display the first few rows of 'Payment_Type' and 'payment_method' columns
```

```python
df.groupby('Payment_Type').size()    # Count the number of orders for each payment type
```

```python
def revenue_category(row):
    if row['revenue'] > 500:
        return 'High'
    elif row['revenue'] >= 200:
        return 'Medium'
    else:
        return 'Low'

df['Revenue_Category'] = df.apply(revenue_category, axis=1)  # Apply the revenue_category function to each row
```

```python
df['Revenue_Category'].value_counts()  # Count the number of orders in each revenue category
```

```python
df[['revenue', 'Revenue_Category']].sample(10)  # Display the first few rows of 'revenue' and 'Revenue_Category' columns
```

```python
# pipe() - method chaining
def add_profit(data):
    data['Profit'] = data['revenue'] * 0.20 # Add a new column 'Profit' as 20% of 'revenue'
    return data

def add_net_revenue(data):
    data['Net_Revenue'] = data['revenue'] - data['Profit'] # Add a new column 'Net_Revenue' as 'revenue' minus 'Profit'
    return data

df = (
    df
    .pipe(add_profit)  # Apply the add_profit function
    .pipe(add_net_revenue)  # Apply the add_net_revenue function
)
```

```python
df[["revenue", "Profit", "Net_Revenue"]].head()  # Display the first few rows of 'revenue', 'Profit', and 'Net_Revenue' columns
```

Topic 66 | Working with Datetime & Time Series Data
This topic covers parsing dates, setting datetime indexes, and resampling data across time periods. It highlights computing monthly totals and growth rates.

```python
import pandas as pd
df = pd.read_csv("ecommerce_sales_analytics_5000.csv", parse_dates=["order_date"])
df.info()  # Print information about the DataFrame
```

```python
# Setting datetime index
df = df.set_index('order_date')  # Set 'order_date' as the index of the DataFrame
```

```python
df.index
```

```python
#Monthly Revenue Analysis
monthly_revenue = df['revenue'].resample('ME').sum()  # Resample the 'revenue' column by month and calculate the sum
monthly_revenue
```

```python
# Trend interpretation
monthly_revenue.pct_change()  # Calculate the percentage change in monthly revenue
```

```python
monthly_revenue.index.is_monotonic_increasing  # Check if the index is monotonically increasing
```

```python
print(monthly_revenue.sum(), df['revenue'].sum())  # Print the total monthly revenue and the total revenue from the original DataFrame
monthly_revenue.sum() <= df['revenue'].sum()  # Check if the total monthly revenue is less than or equal to the total revenue from the original DataFrame   
```
