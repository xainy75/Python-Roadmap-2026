```python
for char in "Python":               #iterating through a string
    print(char)
```

```python
i = 4
rows = i
cols = i

matrix = []
for r in range(rows):
    row = []
for c in range(cols):
    cols = []
    row.append(0)
    matrix.append(row)

for row in matrix:
    print(' '.join(map(str, row)))
```

Topic 20 | Nested Loops in Python

This topic covers nested loops to handle multi‐level operations. Learners gain the ability to solve more complex problems.

```python
for i in range(3):              #nested loop
    print("Outer Loop")
    for j in range(2):
        print("     Inner Loop")
```

```python
i = 4
rows = i
cols = i

matrix = []
for r in range(rows):
    row = []
    for c in range(cols):
        row.append(0)  # Append a default value (e.g., 0) to the row
    matrix.append(row)

for row in matrix:
    print(' '.join(map(str, row)))
```

Topic 26 | Lambda Functions in Python
This topic covers anonymous lambda functions and how they simplify small tasks. Learners also use lambda with map() and filter().

```python
add = lambda a, b: a + b
print(add(3, 5))
```

```python
double_it = lambda n: n * 2
print(double_it(5))
```

```python
num = [1, 2, 3, 4, 5]
squares = list(map(lambda x: x * x, num))
print(squares)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
odd_numbers = list(filter(lambda x: x % 2 != 0, numbers))
print(odd_numbers)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)
```
