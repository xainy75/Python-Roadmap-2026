Task 1: Dataset Loading

```python
import pandas as pd

# load dataset
df = pd.read_csv(r"C:\Users\User\Downloads\HousePricePrediction.csv")
# display first 10 rows of the dataset
print(df.head(10))
```

Task 2: Data Exploration

```python
# display shape of dataset
print(df.shape)

# Check data types of all columns
print(df.dtypes)

#Generate summary statistics of numerical features
print(df.describe())
```

Task 3: Data Cleaning

```python
#Identify missing values
print(df.isnull().sum())

#Handle missing values appropriately
df = df.fillna(df.mean(numeric_only=True))

# Remove duplicate records
df = df.drop_duplicates()

# Verify duplicates removed
print("\nDuplicate Records:")
print(df.duplicated().sum())
```

Task 4: Feature Selection

```python
#Identify relevant features for analysis
if 'ID' in df.columns:
    df = df.drop('ID', axis=1)

# Display remaining columns
print(df.columns)
```

Task 5: Data Preprocessing

```python
# Convert categorical variables into numerical form
df_encoded = pd.get_dummies(df)

# Display preview of processed dataset
print(df_encoded.head())
```
