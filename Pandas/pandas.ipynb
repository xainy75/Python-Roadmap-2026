```python
import pandas as pd
df = pd.read_csv("C:\\Users\\User\\Downloads\\Digiskill Python files\\retail_sales.csv")
x = df["Sales"].mean()
print(x)
df.fillna({"Sales": x}, inplace = True) # to fill the missing values in the "Sales" column with the mean value of that column
print(df.to_string())
y = df["Quantity"].mean()
print(y)
df.fillna({"Quantity": y}, inplace = True)
print(df.to_string())
z = df["Profit"].mean()
print(z)
df.fillna({"Profit": z}, inplace = True)
print(df.to_string())


```

```python
import pandas as pd
df = pd.read_csv("C:\\Users\\User\\Downloads\\Digiskill Python files\\retail_sales.csv")
for col in df.select_dtypes(include="number").columns:
    df[col] = df[col].fillna(df[col].mean())
print(df.to_string())
```

```python
df[df["Sales"]>1000]   # to filter the data based on a condition
print(df[df["Sales"]>1000].to_string())
```

```python
df["Total"] = df["Sales"] + df["Profit"]
print(df.to_string())
```

```python
df["Revenue"] = df["Sales"] - df["Profit"]
print(df.to_string())
```

```python
df["Profit Margin"] = df["Profit"] / df["Sales"] * 100
print(df.to_string())
```

```python
df["Date"] = pd.to_datetime(df["Date"], format = "mixed")
print(df["Date"].dtype)  # should now show datetime64[ns]
```

```python
df.dropna(subset=["Date"], inplace=True)  # remove rows where "Date" is NaT
print(df["Date"].dtype)  # should now show datetime64[ns]
```

```python
print(df["Date"].dtype)  # should now show datetime64[ns]
```

```python
df["Date"] = pd.to_datetime(df["Date"], format="%m/%d/%Y", errors="coerce")
print(df[df["Date"].isna()])
df = df.dropna(subset=["Date"])
df["Month"] = df["Date"].dt.month_name()
df["Year"] = df["Date"].dt.year
print(df["Date"].dtype)
print(df.to_string())
```
