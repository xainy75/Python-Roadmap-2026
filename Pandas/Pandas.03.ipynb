```python
df.dropna(inplace=True)   # drop all the rows with missing values and update the original DataFrame
print(df.info())
```

```python
mask = df.replace([r'^\s*$', np.nan], regex=True).isna().any(axis=1) # replace all the empty strings with NaN values
print(mask)
```

```python
# remove the column form data
df = df.drop(columns = 'Unnamed: 22')
```

```python
df.info()
```

```python
df["Total Revenue"] = df["Amount"].sum()
print(df.info())
```

```python
df["Total Order"] = len(df["Order ID"].unique())
print(df.info())
```

```python
df.describe()   # find the summary statistics of the DataFrame
```

```python
df["Average Order Value"] = df["Total Revenue"] / df["Total Order"]
print(df.info())
```

```python
df["Cancelled Orders"] = df[df["Courier Status"] == "Cancelled"].shape[0]
df["Cancelled Rate"] = df["Cancelled Orders"] / df["Total Order"]
df.info()
```

```python
df["Delivered Orders"] = df[df["Courier Status"] == "Delivered"].shape[0]
df["Delivery Rate"] = df["Delivered Orders"] / df["Total Order"]
df.info()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')   # plot a heatmap of the correlation matrix
plt.show()
```
