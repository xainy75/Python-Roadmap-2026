```python
My_df = df.dropna()  # drop rows with missing values
print(My_df.info())   # check for missing values again
print(My_df.head())
```

```python
df.dropna(inplace=True)  # drop rows with missing values in place
print(df.info())         # check for missing values again
print(df.to_string())  # print entire DataFrame
```

```python
df["Debt_to_Income_Ratio"] = df["loan_amount"] / df["income"]
print(df.to_string())  # print entire DataFrame with new column
```

```python
df["Remaining_Debt"] = df["income"] - df["loan_amount"]
print(df.to_string())  # print entire DataFrame with new column
```

```python
bins = [0, 30000, 60000, 90000, 120000]
labels = ["Low", "Medium", "High", "Very High"]
df["Income_Level"] = pd.cut(df["income"], bins=bins, labels=labels)
print(df.to_string())  # print entire DataFrame with new column
```

```python
max =df["income"].max()
min = df["income"].min()
mean = df["income"].mean()
print(f"Max: {max}, Min: {min}, Mean: {mean}")
```
